```python
import random

battery = 100
actions = ["vacuuming", "mopping", "dusting"]

while battery > 0:
    current_action = random.choice(actions)
    print(f"The robot is {current_action}. (Battery: {battery}%)")

    # 1. Subtract a random amount from the battery
    battery -= random.randint(5, 15)

    # 2. Check if battery is low
    if battery <= 20:
        # Randomly decide if a charging station is nearby
        if random.choice([True, False]):
            print("🔋 Battery is low, but found a charging dock! Recharging to 100%.")
            battery = 100
        else:
            print("🪫 Battery is critical and no dock is nearby. Shutting down...")
            break

print("Robot is powered off.")
```
